```python
# [CELL 1] PYTHON RUNTIME TEST
import sys
import platform

print("Python Version:", sys.version.split()[0])
print("Operating System:", platform.system())
print("Python Executable:", sys.executable)
```

```python
# [CELL 1] STUDENT CONFIGURATION
LAST_NAME = "Segundo"
STUDENT_ID = "TUPM-26-1952"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
# [CELL 2] LARGEST-VALUE ALGORITHM
data = [
    10 + SEED_DIGIT,
    5 + NAME_LENGTH,
    20 + SEED_DIGIT,
    8 + NAME_LENGTH,
    15 + ID_SUM
]

largest = data[0]
comparisons = 0

for value in data[1:]:
    comparisons += 1
    if value > largest:
        largest = value

print("Data:", data)
print("Largest Value:", largest)
print("Comparisons:", comparisons)
```

```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

sizes = [10, 100, 1000, 5000, 10000]

for n in sizes:
    print(f"n = {n:5d} | operations = {linear_operations(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | 0(1)=1 | "
        f"0(logn)={log_ops(n):2d} | "
        f"0(n)={n:4d} | 0(n²)={n*n:7d}"
    )
```
