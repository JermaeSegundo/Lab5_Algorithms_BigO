```python
# ==========================================
# EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH
# ==========================================

# Student Configuration Inputs
LAST_NAME = "SEGUNDO"
STUDENT_ID = "TUPM-26-1952"
FAVORITE_ARTIST = "LANY"

# Compute Student-Specific Derived Variables
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 25
NAME_LENGTH = len(LAST_NAME)  # 7
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 4
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())  # 52

# Data Generation Rules
# Generate original sorted incident records dataset
original_dataset = [100 + i * 5 + SEED_DIGIT for i in range(12)]

# Determine rotation point using SEED_DIGIT
pivot = (SEED_DIGIT % len(original_dataset)) + 1
rotated_dataset = original_dataset[pivot:] + original_dataset[:pivot]

# Generate target incident using ID_SUM and ARTIST_SUM
target_incident = original_dataset[(ID_SUM + ARTIST_SUM) % len(original_dataset)]

# Linear Search Implementation
def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

# Rotated Binary Search Implementation (Lab 5 Algorithm Concept)
def binary_search_rotated(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid, comparisons
            
        # Check if the left half is sorted
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Otherwise, the right half is sorted
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

# Execute Search Algorithms
lin_index, lin_comparisons = linear_search(rotated_dataset, target_incident)
bin_index, bin_comparisons = binary_search_rotated(rotated_dataset, target_incident)

# 5. Assessment Data & Output Summary
print("========== EXERCISE 1 ASSESSMENT DATA ==========")
print(f"Original Dataset      : {original_dataset}")
print(f"Rotated Dataset       : {rotated_dataset}")
print(f"Target Incident       : {target_incident}")
print("-" * 50)
print(f"Rotated Binary Search : Index = {bin_index} | Comparisons = {bin_comparisons}")
print(f"Linear Search         : Index = {lin_index} | Comparisons = {lin_comparisons}")
print("-" * 50)
print("Time and Space Complexity Analysis:")
print("  - Rotated Binary Search : O(log n) Time | O(1) Auxiliary Space")
print("  - Linear Search         : O(n) Time | O(1) Auxiliary Space")
print(f"\nFinal Output: Target incident {target_incident} successfully located at index {bin_index} using {bin_comparisons} comparisons.")
```
