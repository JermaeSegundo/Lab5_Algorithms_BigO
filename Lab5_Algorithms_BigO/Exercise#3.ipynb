```python
# ==========================================
# EXERCISE 3: E-COMMERCE ORDER PROCESSING
# ==========================================

import random
import time

# 1. Student Configuration Inputs
LAST_NAME = "SEGUNDO"
STUDENT_ID = "TUPM-26-1952"
FAVORITE_ARTIST = "LANY"

# Compute Student-Specific Derived Variables
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 25

# Data Generation Rules (Base Orders Dataset)
random.seed(SEED_DIGIT)
base_orders = [(ID_SUM * 4 + i * 5 + random.randint(1, 15)) for i in range(12)]

# Define the Three Input Conditions Required by Lab 5
input_conditions = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}

# Sorting Algorithms Implementation (Tracking Comparisons and Movements)
def bubble_sort(arr):
    a = arr.copy()
    comps, moves = 0, 0
    n = len(a)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comps += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                moves += 1
                swapped = True
        if not swapped:
            break
    return a, comps, moves

def selection_sort(arr):
    a = arr.copy()
    comps, moves = 0, 0
    n = len(a)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            comps += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            moves += 1
    return a, comps, moves

def insertion_sort(arr):
    a = arr.copy()
    comps, moves = 0, 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if a[j] > key:
                a[j + 1] = a[j]
                moves += 1
                j -= 1
            else:
                break
        a[j + 1] = key
    return a, comps, moves

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

# Execution & Benchmarking across all input conditions
print("========== EXERCISE 3 ASSESSMENT DATA ==========")
print(f"Generated Base Orders Dataset : {base_orders}\n")

for cond_name, data in input_conditions.items():
    print(f"--- Condition: {cond_name} ---")
    print(f"{'Algorithm':<15} {'Comparisons':<12} {'Movements':<12} {'Time (s)':<15} {'Verification'}")
    print("-" * 68)
    
    for alg_name, func in algorithms.items():
        start_time = time.perf_counter()
        sorted_arr, comparisons, movements = func(data)
        elapsed_time = time.perf_counter() - start_time
        
        is_verified = "Passed" if sorted_arr == sorted(data) else "Failed"
        print(f"{alg_name:<15} {comparisons:<12} {movements:<12} {elapsed_time:<15.6f} {is_verified}")
    print()

# Big O & Case Analysis Summary
print("Big O Analysis & Case Observations:")
print("  - Bubble Sort    : Best O(n) on Sorted data (with early exit) | Worst O(n^2) on Reverse-Ordered data.")
print("  - Selection Sort : O(n^2) fixed comparisons across all conditions regardless of initial order.")
print("  - Insertion Sort : Best O(n) with 0 shifts on Sorted data | Worst O(n^2) on Reverse-Ordered data.")
print("\nFinal Output: All three sorting algorithms successfully tested and verified across all input conditions.")
```
