```python
# ==========================================
# EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS
# ==========================================

# Student Configuration Inputs
LAST_NAME = "SEGUNDO"
STUDENT_ID = "TUPM-26-1952"
FAVORITE_ARTIST = "LANY"

# Compute Student-Specific Derived Variables
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 25
NAME_LENGTH = len(LAST_NAME)  # 7
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 4

# Data Generation Rules
# Dataset A (Sorted ascending) using SEED_DIGIT and ID_SUM
dataset_a = [1000 + i * (SEED_DIGIT + 3) + ID_SUM for i in range(5)]

# Dataset B (Sorted ascending) using ARTIST_LENGTH and NAME_LENGTH
dataset_b = [1001 + i * (ARTIST_LENGTH + 2) + NAME_LENGTH for i in range(6)]

# Two-Pointer Consolidation (Merge) Implementation
def consolidate_datasets(arr1, arr2):
    i = 0
    j = 0
    consolidated = []
    comparisons = 0
    
    # Compare elements from both arrays and merge in ascending order
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            consolidated.append(arr1[i])
            i += 1
        else:
            consolidated.append(arr2[j])
            j += 1
            
    # Append any remaining elements from Dataset A
    while i < len(arr1):
        consolidated.append(arr1[i])
        i += 1
        
    # Append any remaining elements from Dataset B
    while j < len(arr2):
        consolidated.append(arr2[j])
        j += 1
        
    return consolidated, comparisons

# Execute Consolidation
consolidated_dataset, total_comparisons = consolidate_datasets(dataset_a, dataset_b)
is_correctly_sorted = consolidated_dataset == sorted(consolidated_dataset)

# Assessment Data & Output Summary
print("========== EXERCISE 2 ASSESSMENT DATA ==========")
print(f"Department A Dataset : {dataset_a}")
print(f"Department B Dataset : {dataset_b}")
print(f"Consolidated Dataset : {consolidated_dataset}")
print("-" * 50)
print(f"Records Before Merge : {len(dataset_a) + len(dataset_b)} total records")
print(f"Records After Merge  : {len(consolidated_dataset)} total records")
print(f"Comparison Count     : {total_comparisons}")
print(f"Ordering Verification: {'Passed (Successfully Sorted)' if is_correctly_sorted else 'Failed'}")
print("-" * 50)
print("Time and Space Complexity Analysis:")
print(f"  - Time Complexity  : O(n + m) where n={len(dataset_a)}, m={len(dataset_b)}")
print("  - Space Complexity : O(n + m) auxiliary space for the consolidated collection")
print(f"\nFinal Output: Successfully consolidated into {len(consolidated_dataset)} ordered enrollment records.")
```
